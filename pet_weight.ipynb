```python
PetAge_Years=pd.DataFrame(df['PetAge_Years'])
PetWeight_kg=pd.DataFrame(df['PetWeight_kg'])
```

```python
from sklearn import linear_model
```

```python
lm=linear_model.LinearRegression()
model=lm.fit(PetAge_Years,PetWeight_kg)
```

```python
new_PetAge_Years=([[9.25]])
new_PetWeight_kg=model.predict(new_PetAge_Years)
print("Predicted score for 9.25 hours study/days=",new_PetWeight_kg)
```
